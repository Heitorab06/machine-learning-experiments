```python
def evaluate_model(model, X, y):

    prediction = model.predict(X)

    metrics = {
        "cccuracy": accuracy_score(y, prediction),
        "precision": precision_score(y, prediction, average="macro"),
        "recall": recall_score(y, prediction, average="macro"),
        "f1": f1_score(y, prediction, average="macro")
    }
    print("Accuracy:", metrics["accuracy"])
    print("Precision:", metrics["precision"])
    print("Recall:", metrics["recall"])
    print("F1:", metrics["f1"])

    return metrics
```

```python
from pathlib import Path

RESULTS_FILE =  Path("experiments_history.csv")

if RESULTS_FILE.exists():
    results_df = pd.read_csv(RESULTS_FILE)
else:
    results_df = pd.DataFrame(columns=[
    "model",
    "params",
    "accuracy",
    "precision",
    "recall",
    "f1"
])
    results_df.to_csv(RESULTS_FILE, index=False)
```

```python
from typing import Any

def save_result(model_name, params: dict[str, Any], metrics: dict[str, float]):
    global results_df

    new_result = pd.DataFrame([{
    "model": model_name,
    "params": params,
    **metrics
}])

    results_df = pd.concat([results_df, new_result], ignore_index=True)
    results_df.to_csv(RESULTS_FILE, index=False)
```
